### Tracking Shared State

Counting how many objects have been created is a classic reason to promote a property from the instance level to the class level. We first show the brittle `global` approach, then migrate to proper class variables and class methods.


```python
# add functionality to the class ?
class Student:
    def __init__(self, name='', email='', salary=100 ):
        self.name = name
        self.email =email
        self.salary = salary

    # instance method ?? --> behaviour depends on caller instance
    def displayStudent(self):
        # first param is self ?? # represent current calling object
        print(f'name={self.name}, {self.email}')

ss = Student("Ahmed", "ahmed@gmail.com",2324)
ss.displayStudent()
```

```python
# count number of objects created from a class
no_of_objects = 0
class Employee:
    def __init__(self, name, salary):
        global no_of_objects
        no_of_objects +=1
        self.name = name
        self.salary = salary


    def displayInfo(self):
        print(f"Employee name={self.name}")

emp = Employee('noha', 13124525)
emp2 = Employee("mmm", 1234124)
print(no_of_objects)
print(emp)
```

### Class Methods as Factories

Re-using the `Employee` class, we now teach it how to spawn fully configured instances via `@classmethod`. Notice how the method receives the class (`cls`) and can therefore decide which subclass to instantiate.


```python
# use class method --> apply functionality ... related to the class

# object creation >>>
class Employee:
    no_of_objects = 0 # class variable ?
    nationality = "Egyptian"

    def __init__(self, name, salary):

        self.name = name
        self.salary = salary
        Employee.no_of_objects +=1

    def displayInfo(self):
        print(f"Employee name={self.name}")

    # create object with default values from class ?
    @classmethod
    # first parm of function represnet the class it self
    # Usually first parameter of such methods is named 'cls'
    def create_default_object(abbbass):
        print(abbbass)
        return  abbbass("default", 2324434)

emp = Employee.create_default_object()
print(emp, emp.__dict__)
```

```python
# use class method --> apply functionality ... related to the class

# object creation >>>
class Employee:
    no_of_objects = 0 # class variable ?
    nationality = "Egyptian"

    def __init__(self, name, salary):

        self.name = name
        self.salary = salary
        # Employee.no_of_objects +=1
        self.__class__.no_of_objects +=1

    def displayInfo(self):
        print(f"Employee name={self.name}, iti")

    # create object with default values from class ?
    # class methods  ==> object factory
    @classmethod
    # first parm of function represnet the class it self
    # Usually first parameter of such methods is named 'cls'
    def create_default_object(cls):
        print(cls)
        return  cls("default", 2324434)

emp = Employee.create_default_object()
emp.displayInfo()
print(emp, emp.__dict__)
```

## 4. Inheritance Preview

To cap things off we stub out a `Person` base class and derive `Student` from it. Future exercises will flesh this relationship out with overridden methods and shared utilities.


```python


class Employee:
    no_of_objects = 0 # class variable ?
    nationality = "Egyptian"

    def __init__(self, name,salary):
        self.name = name
        self.salary = salary
        # Employee.no_of_objects +=1
        self.__class__.no_of_objects +=1

    def displayInfo(self):
        print(f"Employee name={self.name}, iti")

    @classmethod
    def create_default_object(cls):
        print(cls)
        return  cls("default", 2324434)

    # this is a helpler function , doesn't depend on the instance or the class
    @staticmethod  # used with helper method
    def cal_tax(salary):
        print(no_of_objects)
        # salary is value passed to the function
        print(salary)
        return  salary*.8

emp = Employee("noha",1000)
emp.cal_tax(1000)
Employee.cal_tax(4000)
Employee.cal_tax(emp.salary)
```
