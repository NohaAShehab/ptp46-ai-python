- 📖 Reading from files (different methods)

## 🎯 Basic File Operations

The **three essential steps** for file operations:

1. **🔓 Open** the file
2. **⚙️ Perform** operations (read/write)
3. **🔒 Close** the file

### 📋 File Modes:

| Mode | Description | Behavior |
|------|-------------|----------|
| `r` | **Read** mode (default) | Opens file for reading |
| `w` | **Write** mode | Overwrites existing content |
| `a` | **Append** mode | Adds to end of file |

> **Remember:** Always close files to free up system resources!


```python
# read, write data to text file .

# 1- open file
"""
 open (filename, mode), default mode= r
 mode : r- > read , w -> write, a -> append
"""
fileobj = open("mycv.")
# 2- do operation
print(fileobj)
# 3- close
fileobj.close() # close connection
```

---

## 🛡️ Error Handling for File Operations

Always use **`try-except` blocks** when working with files to handle cases where:

- ❌ The file doesn't exist
- 🔒 Permission errors occur
- ⚠️ Other I/O errors happen

> **Best Practice:** Always wrap file operations in try-except to prevent crashes!


```python
try:
    fileobj = open("mycv", "r") # if file not exists --> raise error
except Exception as e:
    print(e)
else:
    # 2- do operation
    print(fileobj)
    # 3- close
    fileobj.close()
```

---

## 📖 Reading Files: `read()` Method

The `read()` method reads the **entire file content as a single string**. 

**Use Cases:**
- ✅ Small files
- ✅ When you need all content at once
- ✅ Simple text processing

> **⚠️ Warning:** Not recommended for large files as it loads everything into memory!


```python
try:
    fileobj = open("mycv", "r") # if file not exists --> raise error
except Exception as e:
    print(e)
else:
    # 2- do operation
    print(fileobj)
    data = fileobj.read() # read content into one string
    print(data, type(data))
    # 3- close
    fileobj.close()
```

---

## 📋 Reading Files: `readlines()` Method

The `readlines()` method reads **all lines from the file and returns them as a list**. 

**Characteristics:**
- 📝 Each line includes the newline character (`\n`) at the end
- 📊 Returns a list of strings
- 🔄 Useful when you need to process lines individually

> **Tip:** Use list comprehension with `readlines()` for filtering: `[line.strip() for line in file.readlines()]`


```python
try:
    fileobj = open("mycv", "r") # if file not exists --> raise error
except Exception as e:
    print(e)
else:
    # 2- do operation
    print(fileobj)
    data = fileobj.readlines() # read content into a list, each item line
    # line ending with \n
    print(data, type(data))
    # 3- close
    fileobj.close()
```

---

## 🔄 Reading Files: Iterating Line by Line

You can **iterate directly over the file object** to read line by line. 

**Advantages:**
- 💾 **Memory-efficient** for large files
- ⚡ Processes one line at a time
- 🎯 Best practice for large datasets

> **⚠️ Important:** After iterating through the file, the file pointer is at the end. Calling `read()` will return an empty string!


```python
try:
    fileobj = open("mycv", "r") # if file not exists --> raise error
except Exception as e:
    print(e)
else:
    for line in fileobj: # catch line by line from the file.
        print(line)
    print(fileobj.read(), "content ")
    # 3- close
    fileobj.close()
```

---

## 🔙 File Pointer: `seek()` Method

The `seek(0)` method **resets the file pointer** to the beginning of the file. 

**Use Cases:**
- 🔄 Re-reading a file after iteration
- 📍 Moving to a specific position in the file
- 🔁 Multiple read operations on the same file

> **Syntax:** `file.seek(offset)` - `offset` is the byte position (0 = beginning)


```python
try:
    fileobj = open("mycv", "r") # if file not exists --> raise error
except Exception as e:
    print(e)
else:
    for line in fileobj: # catch line by line from the file.
        print(line)
    # reset file pointer to the beginning of the file
    fileobj.seek(0)
    print(fileobj.read(), "content ")
    # 3- close
    fileobj.close()
```

## ✍️ Writing to Files

### 📝 Write Mode (`w`)

When opening a file in write mode (`w`):

| Scenario | Behavior |
|----------|----------|
| 📄 File **doesn't exist** | ✅ File will be **created** |
| 📄 File **exists** | ⚠️ File will be **opened and truncated** (all existing content removed) |
| 📍 Writing position | Starts from the **beginning** of the file |

**Usage:**
- Use `write()` method to write strings to the file
- Remember to add `\n` for new lines
- Multiple `write()` calls append to the current position

> **⚠️ Warning:** Write mode **overwrites** existing content. Use append mode (`a`) to preserve data!

## ➕ Appending to Files

### 📎 Append Mode (`a`)

When opening a file in append mode (`a`):

| Scenario | Behavior |
|----------|----------|
| 📄 File **doesn't exist** | ✅ File will be **created** |
| 📄 File **exists** | ➕ New content will be **added to the end** |
| 📝 Existing content | ✅ **Preserved** (not overwritten) |

**Use Cases:**
- 📊 **Logging** - Add log entries without losing history
- 📝 **Data collection** - Append new records to existing data
- 🔄 **Continuous updates** - Add information incrementally

> **💡 Tip:** Append mode is perfect for maintaining data history and building up files over time!


```python
try:
    fileobj = open("users.txt", "w") # if file not exists --> will create it ?? if file exists --> open file for writing starting from the begining of the file  ==> remove old content
    # permissions
except Exception as e:
    print(e)
else:
    print(fileobj)
    fileobj.write("************************\n")
    fileobj.write("We love the lecture\n")
    fileobj.write("we don't need any breaks\n")
    fileobj.write("ITI <3 <3")
    # 3- close
    fileobj.close()
```
