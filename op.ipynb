## Customizing Constructors
The next group of cells experiments with child constructors: try running each one to see what happens when you forget required arguments, add new parameters, or call `super().__init__()` to keep the parent state intact.


```python
# customize employee
class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    def __init__(self, salary):
        self.salary = salary

emp = Employee()
```

```python
class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    def __init__(self, salary):
        self.salary = salary

emp = Employee(23534646)
```

```python
# to call parent constructor you must call it explicityly

class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    # override  parent construcotr
    def __init__(self, name, salary):
        # super call parent constructor .
        super().__init__(name)
        self.salary = salary

emp = Employee("noha", 23235)
```

## Method Overriding
The following trio of cells shows progressive overrides. Start with inheriting the base behavior, then replace it entirely, and finally call `super()` to extend the parent implementation.


```python
class Grand:
   def __init__(self, name="name"):
       self.name=  name
       print("Grand is called ")



class Parent1(Grand):
   def __init__(self, salary=100):
       super().__init__()
       self.salary = salary
       print('---- parent01 is called')


class Parent2(Grand):
     def __init__(self, email="email"):
        self.email = email
        print('---- parent02 is called')


class Child(Parent1, Parent2):
    pass


c = Child()
print(c.__dict__)
```

```python
class Grand:
   def __init__(self, name="name"):
       self.name=  name
       print("Grand is called ")



class Parent1(Grand):
   def __init__(self, salary=100):
       super().__init__()
       self.salary = salary
       print('---- parent01 is called')


class Parent2(Grand):
     def __init__(self, email="email"):
         super().__init__()
         self.email = email
         print('---- parent02 is called')


class Child(Parent1, Parent2):
    pass


c = Child()
print(c.__dict__)
```

```python
# overriding
class User:
    def sayHello(self):
        print("Hello from user")

class Student(User):
    pass

s = Student()
s.sayHello()
```

## Method Overloading (or the Lack Thereof)
Python keeps only the most recently defined function with a given name. Run the next cells to see how later definitions shadow earlier ones and why decorators like `@overload` from `typing` are used for documentation rather than runtime behavior.


```python
# overriding
class User:
    def sayHello(self):
        print("Hello from user")

class Student(User):
    def sayHello(self):
        print("Hello from student")

s = Student()
s.sayHello()
```

```python
# overriding
class User:
    def sayHello(self):
        print("Hello from user")

class Student(User):
    def sayHello(self):
        super().sayHello()
        print("Hello from student")

s = Student()
s.sayHello()
```

```python
# overloading
class Car:
    def create_car(self):
        print("-- creating new car")

    def create_car(self, color):
        print(f'car created with color {color}')

cc = Car()
cc.create_car()
```

```python
# no explicit overloading in python
# https://typing.python.org/en/latest/spec/overload.html
```

### Quick Recap
- Call `super().__init__()` whenever a child class should inherit parent state.
- Use diagnostic prints or inspect `Child.__mro__` to understand multiple inheritance resolution.
- Override methods to replace behavior; call `super()` inside the override to extend it.
- Python lacks runtime overloading, so prefer default values, keyword-only arguments, or helpers like `functools.singledispatch`.
