```python
# slice string
newmessage= message[3:5]
print(newmessage)
```

```python
### Advanced Slicing with Step
"""
You can use a step value in slicing: `[start:end:step]`
- `[::2]` - every second character
- `[::-1]` - reverse the string'
"""
```

```python
print(message[::2])
```

```python
print(message[::-1]) # return with new string
```

```python

print(message[::-1]) # return with new string
```

## ✨ String Formatting {#formatting}

Python provides several ways to format strings:
- **String concatenation** with `+`
- **`.format()` method** with positional or keyword arguments
- **f-strings** (formatted string literals) - the modern and recommended approach

```python
# define template ==> could be used later ??
temp = "We have {0} students in {1} course"
# standalone string
print(temp)
```

## 🧹 Stripping White Spaces {#stripping}

The `strip()` method removes whitespace (or specified characters) from the beginning and end of a string. Python also provides `lstrip()` and `rstrip()` for one-sided stripping.

```python
print(msg3.strip("@ &$y"))
```

```python
msg = "      we are very happy                    "
print(len(msg))
```

```python
#
txt  =" <h1>Hello string</h1>  "
print(txt.strip("<h1>/"))
```

```python
msg2 = msg.strip() # strip white spaces from the begining and the end of the string
print(msg2, len(msg2))
```
