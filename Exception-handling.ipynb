## 📝 Capturing Exception Details

You can capture the exception object to get more information about what went wrong:

```python
except Exception as e:
    print(f"Error: {e}")
```

The exception object `e` contains details about the error that occurred.


````python
try:
    # try this
    age = int(input("please enter age: "))
except:
    # if any runtime error happened, ## do this
    print("-- Error happened ")
    # save state, save point I am working on ....
    print("-- You entered invalud age, so age = 0 ")
    # I need to handle the exception
    age = 0
    exit(0)
````

## 🎯 Handling Multiple Exception Types

You can handle different types of exceptions separately:

```python
try:
    # code
except ZeroDivisionError:
    # Handle division by zero
except ValueError:
    # Handle invalid value
except Exception:
    # Handle any other exception
```

**Common Exception Types:**
- `ZeroDivisionError` - Division by zero
- `ValueError` - Invalid value type
- `TypeError` - Wrong type used
- `FileNotFoundError` - File doesn't exist


````python
try:
    num1 = int(input("please enter first number "))
    num2 = int(input("please enter second number"))
    res = num1/num2
    print(res)
except :
    print("--- error happened ")
````

## ✅ The `else` Clause

The `else` block executes **only if no exceptions occur** in the `try` block:

```python
try:
    # risky code
except Exception:
    # handle error
else:
    # executed only if no exception occurred
```

This is useful for code that should only run when the operation succeeds.


````python
try:
    num1 = int(input("please enter first number "))
    num2 = int(input("please enter second number"))
    res = num1/num2
    print(res)
    # you may receive exceptions from different types
    # so may need to change how will you handle it .
    # you can add more than one except ?
except Exception as e :  # object contain exception details
    print(f"--- error happened {e}")
````

## 🔒 The `finally` Clause

The `finally` block **always executes**, regardless of whether an exception occurred:

```python
try:
    # risky code
except Exception:
    # handle error
finally:
    # Always executed (cleanup code)
```

**Use cases:**
- Closing files
- Releasing resources
- Cleanup operations
- Logging completion

**Important:** `finally` executes even if there's a `return` statement in `try` or `except`!


````python
try:
    num1 = int(input("please enter first number "))
    num2 = int(input("please enter second number"))
    res = num1/num2
    print(res)
    # you may receive exceptions from different types
    # so may need to change how will you handle it .
    # you can add more than one except ?
except ZeroDivisionError as ze:
    print("-- dividing by zero is not applicable => setting res=0 ")
    res = 0
except ValueError as ve:
    print("cannot convert input string value to integer ")
    res  = None
except Exception as e :
    print(f"--- error happened {e}")
````

## 🚀 Raising Exceptions

You can raise exceptions manually using the `raise` keyword:

```python
if condition:
    raise ExceptionType("Error message")
```

This is useful for:
- Input validation
- Enforcing business rules
- Creating custom error conditions


````python
def sumnum():
    try:
        num1 = int(input("please enter num1:"))
        num2 = int(input("please enter num2: "))
        res = num1/ num2
    except Exception as e:
        print(e,"Error happened " )
        return  False
    else:
        print("===========")
        return res
    print("---- the operation done safelly ")

print(sumnum())
````
