```python
# define template ==> could be used later ??
temp = "We have {0} students in {1} course"
# standalone string
print(temp)
```

```python
# reformat the template with data ??
print(temp.format(22, 'Java'))  # .format # return with new string
```

```python
print(temp.format(no_of_students, coursename))
```

```python
print(temp.format(coursename, no_of_students))
```

<h1>Strip white spaces  </h1>

```python
msg = "      we are very happy                    "
print(len(msg))
```

```python
msg2 = msg.strip() # strip white spaces from the begining and the end of the string
print(msg2, len(msg2))
```

```python
print(msg.lstrip(), len(msg.lstrip()))
```

```python
msg3 = "We are happy          @"
print(msg3.strip("@")) # strip @ only , leave spaces
```
