<h2 style='color: #00BCD4; background: linear-gradient(135deg, #E0F7FA 0%, #B2EBF2 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #00ACC1;'>🔄 Creating a Range Object</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates creating a <code style='background: #F5F5F5; padding: 2px 6px; border-radius: 3px; color: #D32F2F;'>range()</code> object. Range objects are <strong style='color: #E91E63;'>iterable</strong> and memory-efficient for generating sequences of numbers.</p>


```python
# range is iterable object ==>
rr = range(19, 1, -1) # 19, 1 , 1
for i in rr:
    print(i)
```

<h2 style='color: #00BCD4; background: linear-gradient(135deg, #E0F7FA 0%, #B2EBF2 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #00ACC1;'>🔄 Iterating Over Range with Step</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates iterating over a range with a <strong style='color: #E91E63;'>negative step</strong> to count backwards. The range includes start but excludes stop.</p>


```python
# cast range to a list ?

ll = list(range(1,5))
print(ll)
```
