## List Comprehensions

Python lets us squeeze an entire `for` loop and optional filter into a single expression. Below we experiment with three quick variations:
- build a simple range of numbers
- step through the range with a custom stride
- filter values inline with an `if` clause

```python
l = [num for num in range(10)]
print(l)
```

```python
l = [num for num in range(0,10, 2)]
print(l)
```

```python
l = [num for num in range(10) if num%2==0 ]
print(l)
```

## Swapping Without Temp Variables

Tuple unpacking lets us swap two variables in a single line—no placeholders, no helper functions. It is expressive, atomic, and avoids mistakes when juggling multiple assignments.


```python
# sawp
name = 'ali'
track = 'ai'
print(f"name= {name}")
print(f"track= {track}")
print("-----------------------")
name, track = track, name
print(f"name= {name}")
print(f"track= {track}")
```

## Working with Files Safely

`with open(...)` gives us a context manager that closes the file handle automatically—even when an exception bubbles up. Try editing the body to write JSON, CSV, or log lines without worrying about manual cleanup.


```python
with open("users.txt", "w") as fileobj:
    fileobj.write("hello")

print(fileobj)
```
