## 🔁 Variable Scope in Loops

In Python, loop variables **persist after the loop completes**. This demonstrates the scope of the `char` variable:

> **Key Point:** Unlike some languages, Python loop variables remain accessible after the loop finishes.


```python
for char in "iti":
    print(f"char = {char}")

print(f"-- after loop {char}")
```

## 🎭 Function Behavior with Different Types

Python's **dynamic typing** allows functions to work with different data types. The `sumnum` function can:
- ➕ Add numbers
- 🔗 Concatenate strings  
- 📋 Combine lists

> **Note:** The same function works with different types because Python uses duck typing!


```python
def sumnum(num1, num2):
    res = num1 + num2
    print(res)

sumnum(23,44) # sumnum
```

```python
sumnum("Ahmed", "ali") # concat
```

## 📝 Type Hints Introduction

Type hints in Python are used for **documentation purposes**. They don't enforce type checking at runtime, but help with:

- 📚 **Code documentation** - Makes code more readable
- 💡 **IDE autocomplete** - Better code suggestions
- 🔍 **Static type checkers** - Tools like `mypy` can validate types

> **Important:** Type hints are **not enforced at runtime** - they're purely for documentation and tooling!


```python
sumnum([33], ["test"])
```

### 💼 Type Hints in Practice

Even with type hints, Python **still allows different types** to be passed. Notice how the function accepts strings despite the `int` type hint:

> **Example:** A function annotated with `-> int` can still receive strings, lists, or any other type!


```python
sumnum("Ahmed", 10)
```

## we use the type hints for doc. purpose

```python
# We need to validate the params ??

def sumnum(num1:int, num2:int)->int:
    res = num1+ num2
    print(res)
    return res

sumnum(324,423)
```

## ✅ Input Validation

Since type hints **don't enforce runtime type checking**, you must implement validation manually. Here's an example of adding validation to ensure the function receives the correct types:

**Validation Methods:**
- `type()` - Check exact type
- `isinstance()` - Check type (preferred, supports inheritance)

> **Best Practice:** Use `isinstance()` instead of `type()` for better flexibility with inheritance!


```python
sumnum("Ahmed", "Ali") # type hints
```

```python
sumnum("iti", 10)
```

# You must do the validation on your own.

```python
def sumnum(num1 :int, num2: int)->int :
    # add validation rules
    if type(num1)==int and isinstance(num2, int):
        res = num1+ num2
        return  res

    print("------ not valid values")

sumnum(23,323)
sumnum("#2", 34)
```
