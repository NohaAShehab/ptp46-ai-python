<h2 style='color: #FF6B6B; background: linear-gradient(135deg, #FFE5E5 0%, #FFCCCC 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #FF4757;'>📝 Function with Parameters</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell shows how to define a function that <strong style='color: #E91E63;'>accepts arguments</strong> (parameters). Functions require all parameters to be provided when called.</p>


```python
# function accept arguments

def printFullname(firstname,lastname):
    fullname = f"{firstname} {lastname}"
    print(fullname)

printFullname()
```

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell shows a function that both <strong style='color: #E91E63;'>prints and returns</strong> a value. The return value can be formatted (e.g., using <code style='background: #F5F5F5; padding: 2px 6px; border-radius: 3px; color: #D32F2F;'>.title()</code>).</p>

<h2 style='color: #FF6B6B; background: linear-gradient(135deg, #FFE5E5 0%, #FFCCCC 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #FF4757;'>📦 Returning Multiple Values</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates how a function can <strong style='color: #E91E63;'>return multiple values</strong> as a tuple. Python automatically packs multiple return values into a tuple.</p>


```python
def printFullname(firstname,lastname): #firstname is param.
    fullname = f"{firstname} {lastname}"
    print(fullname)
    return fullname.title()


printFullname("rwer", "werwer") #rwer  --> arguments
```

<h2 style='color: #FF6B6B; background: linear-gradient(135deg, #FFE5E5 0%, #FFCCCC 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #FF4757;'>🎯 Functions with Default Parameters</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates <strong style='color: #E91E63;'>optional parameters</strong> with default values. Functions can be called with or without these parameters.</p>


```python
def printFullname(firstname,lastname):
    fullname = f"{firstname} {lastname}"
    print(fullname)
    return fullname.title(), firstname, lastname # tuple


printFullname("rwer", "werwer")
```

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates the <code style='background: #F5F5F5; padding: 2px 6px; border-radius: 3px; color: #D32F2F;'>sep</code> parameter in the <code style='background: #F5F5F5; padding: 2px 6px; border-radius: 3px; color: #D32F2F;'>print()</code> function, which specifies the separator between multiple arguments.</p>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates the <code style='background: #F5F5F5; padding: 2px 6px; border-radius: 3px; color: #D32F2F;'>end</code> parameter in <code style='background: #F5F5F5; padding: 2px 6px; border-radius: 3px; color: #D32F2F;'>print()</code>, which specifies what to print at the end instead of a newline.</p>

<h2 style='color: #FF6B6B; background: linear-gradient(135deg, #FFE5E5 0%, #FFCCCC 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #FF4757;'>➕ Function with Required Parameters</h2>

<h2 style='color: #FF6B6B; background: linear-gradient(135deg, #FFE5E5 0%, #FFCCCC 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #FF4757;'>✅ Correct Default Parameter Position</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell shows the <strong style='color: #4CAF50;'>correct way</strong> to use default parameters - they come at the end. The function can be called with fewer arguments.</p>

<h2 style='color: #FF6B6B; background: linear-gradient(135deg, #FFE5E5 0%, #FFCCCC 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #FF4757;'>🎯 Keyword Arguments</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates calling a function using <strong style='color: #E91E63;'>keyword arguments</strong>, which allows you to specify arguments in any order by name.</p>


```python
def sumnum(num1, num2, num3):
    res = num1 + num2 + num3
    return res

print(sumnum(22,4324,23))
```

<h2 style='color: #FF6B6B; background: linear-gradient(135deg, #FFE5E5 0%, #FFCCCC 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #FF4757;'>📊 Variable Number of Arguments (*args)</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates the <code style='background: #F5F5F5; padding: 2px 6px; border-radius: 3px; color: #D32F2F;'>*args</code> syntax, which allows a function to accept <strong style='color: #E91E63;'>zero or more positional arguments</strong> as a tuple.</p>


```python
def sumnum(num1, num2=10, num3):
    res = num1 + num2 + num3
    return res

print(sumnum(22,4324,23))
```

<h2 style='color: #FF6B6B; background: linear-gradient(135deg, #FFE5E5 0%, #FFCCCC 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #FF4757;'>🔤 String Format Method</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates the <code style='background: #F5F5F5; padding: 2px 6px; border-radius: 3px; color: #D32F2F;'>.format()</code> method for string formatting, which uses positional placeholders.</p>


```python
def sumnum(num1, num2, num3=10):
    print(f'num1={num1}, num2={num2}, num3={num3}')
    res = num1 + num2 + num3
    return res

print(sumnum(22,4324,23))
```

<h2 style='color: #FF6B6B; background: linear-gradient(135deg, #FFE5E5 0%, #FFCCCC 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #FF4757;'>🔑 Variable Number of Keyword Arguments (**kwargs)</h2>

<p style='color: #555; font-size: 14px; line-height: 1.6;'>This cell demonstrates the <code style='background: #F5F5F5; padding: 2px 6px; border-radius: 3px; color: #D32F2F;'>**kwargs</code> syntax, which allows a function to accept <strong style='color: #E91E63;'>zero or more keyword arguments</strong> as a dictionary.</p>


```python
sumnum(num3=100, num2=5, num1=10)
```
