## Guarding State with Getters and Setters
The next implementation introduces validation logic that ensures salaries must be numeric and greater than 1000. It also demonstrates why raising explicit exceptions makes debugging easier.


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name # public
        self._email = email # protected
        self.__salary = salary # private

    def setSalary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary> 1000:
            self.__salary = salary

        else:
            raise  Exception("Salary must number > 1000")

    def getSalary(self):
        return  self.__salary

emp = Employee("test", "test", 12434)
print(emp.getSalary())
emp.setSalary(6000)
print(emp.getSalary())
```

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name # public
        self._email = email # protected
        self.setSalary(salary)

    def setSalary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary> 1000:
            self.__salary = salary
        else:
            raise  Exception("Salary must number > 1000")

    def getSalary(self):
        return  self.__salary

    @property
    def salary(self):
        print("====Salary property is called ")
        return  self.__salary

    @salary.setter
    def salary(self, salary):
        print("-- calling property setter")
        if isinstance(salary, float) or isinstance(salary, int) and salary> 1000:
            self.__salary = salary
        else:
            raise  Exception("Salary must number > 1000")


emppp = Employee("test", "test", 6000)
print(emppp.salary)
emppp.salary= "salary"
```

### Takeaways
- Naming conventions communicate intent even without strict access modifiers.
- Validation logic belongs inside the class so every caller benefits.
- `@property` keeps your interface ergonomic while still enforcing constraints.

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name # public
        self._email = email # protected
        self.salary= salary


    @property
    def salary(self):
        print("====Salary property is called ")
        return  self.__salary

    @salary.setter
    def salary(self, salary):
        print("-- calling property setter")
        if isinstance(salary, float) or isinstance(salary, int) and salary> 1000:
            self.__salary = salary
        else:
            raise  Exception("Salary must number > 1000")
    @salary.deleter
    def salary(self):
        print("salary property is deleted ")
        del self.__salary

emmp = Employee("new", "new", 21124)
del emmp.salary
print(emmp.__dict__)
```
